```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.metrics import mean_squared_error , mean_absolute_error
from sklearn.linear_model import ElasticNet
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
aapl = pd.read_csv("stocks/AAPL.csv")
qqq = pd.read_csv("etfs/QQQ.csv")
msft = pd.read_csv("stocks/MSFT.csv")
nvda = pd.read_csv("stocks/NVDA.csv")
amzn = pd.read_csv("stocks/AMZN.csv")
orcl = pd.read_csv("stocks/ORCL.csv")
googl = pd.read_csv("stocks/GOOGL.csv")
```

```python
aapl["Date"]= pd.to_datetime(aapl["Date"])
aapl.set_index("Date", inplace=True)
print(aapl.shape)
print(aapl.head(5))
```

```python
qqq["Date"]= pd.to_datetime(qqq["Date"])
qqq.set_index("Date", inplace=True)
print(qqq.shape)
print(qqq.head(5))
```

```python
msft["Date"]= pd.to_datetime(msft["Date"])
msft.set_index("Date", inplace=True)
print(msft.shape)
print(msft.head(5))
```

```python
nvda["Date"]= pd.to_datetime(nvda["Date"])
nvda.set_index("Date", inplace=True)
print(nvda.shape)
print(nvda.head(5))
```

```python
amzn["Date"]= pd.to_datetime(amzn["Date"])
amzn.set_index("Date", inplace=True)
print(amzn.shape)
print(amzn.head(5))
```

```python
orcl["Date"]= pd.to_datetime(orcl["Date"])
orcl.set_index("Date", inplace=True)
print(orcl.shape)
print(orcl.head(5))
```

```python
googl["Date"]= pd.to_datetime(googl["Date"])
googl.set_index("Date", inplace=True)
print(googl.shape)
print(googl.head(5))
```

```python
companies={"aapl": aapl,
           "qqq" : qqq,
           "msft": msft,
           "nvda": nvda,
           "amzn": amzn,
           "orcl": orcl,
          "googl": googl}
```

```python
for name, df in companies.items():
    companies[name] = df[df.index >= "2018-01-01"]
```

```python
for name, df in companies.items():
    companies[name] = df["Adj Close"]
```

```python
print(companies["aapl"].shape)
print(companies["qqq"].shape)
print(companies["msft"].shape)
print(companies["nvda"].shape)
print(companies["amzn"].shape)
print(companies["orcl"].shape)
print(companies["googl"].shape)
```

```python
companies["aapl"]
```

```python
returns ={}
for name,df in companies.items():
    returns[name]= df.pct_change()
```

```python
returns["aapl"].head()
```

```python
d_returns=pd.DataFrame(returns)
print(d_returns.head())
print(d_returns.shape)
```

```python
d_returns.dropna(inplace=True)
```

```python
d_returns.isnull().sum()
```

```python
d_returns.shape
```

```python
d_returns["target"] = d_returns["aapl"].shift(-1)
d_returns.shape
```

```python
d_returns.dropna(inplace=True)
```

```python
d_returns.shape
```

```python
train= d_returns[(d_returns.index >= "2018-01-01") & (d_returns.index <"2019-01-01")]
val  = d_returns[(d_returns.index >= "2019-01-01") & (d_returns.index <"2020-01-01")]
test = d_returns[(d_returns.index >= "2020-01-01") & (d_returns.index <"2021-01-01")]
```

```python
print(train.shape)
print(val.shape)
print(test.shape)
```

```python
d_returns
```

```python
x_train= train.iloc[:,:-1]
y_train=train.iloc[:,-1]

x_val= val.iloc[:,:-1]
y_val=val.iloc[:,-1]

x_test= test.iloc[:,:-1]
y_test= test.iloc[:,-1]
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_val.shape)
print(y_val.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
model1= ElasticNet(alpha=0.1, l1_ratio=0.5)
model1.fit(x_train,y_train)
pred_en=model1.predict(x_val)
mae= mean_absolute_error(y_val, pred_en)
rmse =mean_squared_error(y_val, pred_en)**0.5
```

```python
print(mae)
print(rmse)
```

```python
model2= RandomForestRegressor(n_estimators=100)
model2.fit(x_train,y_train)
pred_en2= model2.predict(x_val)
mae2= mean_absolute_error(y_val, pred_en2)
rmse2= mean_squared_error(y_val,pred_en2)**0.5
```

```python
print(mae2)
print(rmse2)
```

```python
model3=GradientBoostingRegressor()
model3.fit(x_train,y_train)
pred_en3=model3.predict(x_val)
mae3= mean_absolute_error(y_val, pred_en3)
rmse3= mean_squared_error(y_val, pred_en3)**0.5 
```

```python
print(mae3)
print(rmse3)
```

```python
pred_ensemble=np.mean([pred_en, pred_en2 , pred_en3], axis=0)
mae4= mean_absolute_error(y_val,pred_ensemble)
rmse4= mean_squared_error(y_val, pred_ensemble)**0.5
```

```python
print(mae4)
print(rmse4)
```

```python
x_train_val= pd.concat([x_train, x_val])
y_train_val= pd.concat([y_train , y_val])

model1.fit(x_train_val,y_train_val)
predict1= model1.predict(x_test)
tmae1= mean_absolute_error(y_test,predict1)
trmse1= mean_squared_error(y_test, predict1)**0.5

model2.fit(x_train_val,y_train_val)
predict2= model2.predict(x_test)
tmae2= mean_absolute_error(y_test,predict2)
trmse2= mean_squared_error(y_test, predict2)**0.5

model3.fit(x_train_val,y_train_val)
predict3= model3.predict(x_test)
tmae3= mean_absolute_error(y_test,predict3)
trmse3= mean_squared_error(y_test, predict3)**0.5

ensemble= np.mean([predict1, predict2, predict3], axis=0)
tmae4= mean_absolute_error(y_test,ensemble)
trmse4= mean_squared_error(y_test, ensemble)**0.5
```

```python
print(tmae1)
print(trmse1)

print(tmae2)
print(trmse2)

print(tmae3)
print(trmse3)

print(tmae4)
print(trmse4)
```

```python
def accuracy(actual_l, pred_l):
    return (np.sign(actual_l)==np.sign(pred_l)).mean()

actual_label= np.sign(y_test)

pred_label1= np.sign(predict1)
model1_accuracy= accuracy(actual_label,pred_label1)

pred_label2= np.sign(predict2)
model2_accuracy= accuracy(actual_label,pred_label2)

pred_label3= np.sign(predict3)
model3_accuracy= accuracy(actual_label,pred_label3)

pred_label4= np.sign(ensemble)
model4_accuracy= accuracy(actual_label,pred_label4)
```

```python
print(model1_accuracy)
print(model2_accuracy)
print(model3_accuracy)
print(model4_accuracy)
```

```python
def pre_hat(value):
    return [value * (1+ i) for i in ensemble]

date_df= x_test.index.to_frame(index=False)
date_df.set_index("Date", inplace=True)
date_df["Ticker"]= ["AAPL" for _ in range (len(date_df)) ]
date_df["r_hat_next"]= ensemble
date_df["p_hat_next"] = companies["aapl"][x_test.index].values * (1 + ensemble)
```

```python
date_df.to_csv("Prediction.csv")
```
